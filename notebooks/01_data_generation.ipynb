# Expense Tracker: Synthetic Data Generation

This notebook demonstrates the raw-data stage of the Expense Tracker App. The records are synthetic and are not real financial facts.

Run it from the project root so imports and project-relative paths work:

```bash
jupyter notebook notebooks/01_data_generation.ipynb
```

The notebook is committed with empty outputs. Run cells from top to bottom and capture reviewed results separately. The configured date window is fixed for reproducibility.


````python
from pathlib import Path
import sys

# The documented launch location is the project root. The fallback helps when
# a notebook server has been opened from inside notebooks/.
PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

from src.config import (
    NUM_RECORDS, DATE_RANGE_DAYS, EXPENSE_CATEGORIES,
    PAYMENT_METHODS, RAW_DATA_PATH, RANDOM_SEED
)
from src.data_generator import DataGenerator

print(f"Project root: {PROJECT_ROOT}")
print(f"Records: {NUM_RECORDS}")
print(f"Configured date range: {DATE_RANGE_DAYS} days")
print(f"Random seed: {RANDOM_SEED}")
print(f"Categories: {len(EXPENSE_CATEGORIES)}")
print(f"Payment methods: {len(PAYMENT_METHODS)}")
````

````python
generator = DataGenerator(num_records=NUM_RECORDS, seed=RANDOM_SEED)
raw_data = generator.generate_data()

print(f"Generated rows: {len(raw_data):,}")
raw_data.head()
````

````python
print("Columns:", list(raw_data.columns))
print("Shape:", raw_data.shape)
print("\nData types:")
print(raw_data.dtypes)
print("\nCategory counts:")
print(raw_data["Category"].value_counts())
print("\nPayment-method counts:")
print(raw_data["Payment_Method"].value_counts())
````

````python
amount_summary = raw_data["Amount"].describe().round(2)
print(amount_summary)
print(f"\nDate range: {raw_data['Date'].min()} to {raw_data['Date'].max()}")
print("\nMissing values:")
print(raw_data.isna().sum())
````

````python
output_path = generator.save_data(RAW_DATA_PATH)
print(f"Saved synthetic raw data to: {output_path}")
print(f"File exists: {Path(output_path).exists()}")
````

````python
category_summary = (
    raw_data.groupby("Category")["Amount"]
    .agg(total="sum", average="mean", count="count")
    .sort_values("total", ascending=False)
    .round(2)
)
category_summary
````
